```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent.parent))
from config1 import CLASSIFICATION_DATA
```

```python
import pandas as pd
import shutil
import os
```

```python
p_img = CLASSIFICATION_DATA / "IMAGES_padded"
p_img_augm = CLASSIFICATION_DATA / "IMAGES_augmented_padded"

p_img_folds =  CLASSIFICATION_DATA / "IMAGES_padded_10folds"
p_img_folds_augm =  CLASSIFICATION_DATA /   "IMAGES_augmDebris_padded_10folds"
```

```python
p_csv = CLASSIFICATION_DATA / "2.MODEL_TRAINING" / "dataset_folds_crossvalidation.csv"

df = pd.read_csv(p_csv)
df
```

```python
# copy images into their corresponding fold folders

def copy_images_to_folds(df, p_img, p_img_folds, augm="no"):
    for line in df.itertuples():
        fold = line.fold_nb
        classes = line.classes
        filename = line.filename
        if augm == "yes":
            filename=filename.replace(".jpg", "_augm.jpg")
        p_img_ini = p_img / classes
        p_img_out = p_img_folds / str(fold) / classes

        os.makedirs(p_img_out, exist_ok=True)
        shutil.copy(p_img_ini / filename, p_img_out / filename)

        
copy_images_to_folds(df, p_img, p_img_folds, augm="no")

copy_images_to_folds(df, p_img_augm, p_img_folds_augm, augm="yes")
```

```python
# check the number of images in the copied fold is correct
c=0
for root, _, files in os.walk(p_img_folds):
    for file in files:
        if file.lower().endswith('.jpg'):
            c+=1
print(c) # 51686 images
```

```python
# check the number of images in the copied fold is correct
# for the dataset of augmented images 
c=0
for root, _, files in os.walk(p_img_folds_augm):
    for file in files:
        if file.lower().endswith('.jpg'):
            c+=1
print(c) # 51686 images
```
