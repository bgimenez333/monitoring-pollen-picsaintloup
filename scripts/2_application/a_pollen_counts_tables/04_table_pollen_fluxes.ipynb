```python
# loading number of Lycopodium spores added in each trap sample for calibration of pollen fluxes
df_ly_tablets = pd.read_csv(p_lyco)
df_ly_tablets = df_ly_tablets[df_ly_tablets["resolution"] == "ANNUEL"]
corresp_ly = {str(df_ly_tablets["site"][i])  + "_" + str(df_ly_tablets["year"][i]) : el for i, el in enumerate(df_ly_tablets["nb_ly_persample"]) }
print(corresp_ly)
df_ly_tablets
```

```python
# Pollen fluxes per traps (without confidence interval)

df_counts["nb_ly_persample"] = [corresp_ly[k] for k in df_counts["site_year"]]

count_ly = {str(df_counts["site"][i])  + "_" + str(df_counts["year"][i]) : float(df_counts["count"][i]) for i, el in enumerate(df_counts["class"]) if el == "Lycopodium"}
print(count_ly)

df_counts["count_ly"] = [count_ly[k] for k in df_counts["site_year"]]

df_counts["count_tot_trap"] = [float(el)*float(df_counts["nb_ly_persample"][e])/float(df_counts["count_ly"][e]) for e, el in enumerate(df_counts["count"])]
df_counts["flux_cm2_an"] = [el / (2.5*2.5*np.pi) for el in df_counts["count_tot_trap"]]

df_counts.head()
```

# with Confidence intervals

```python

df_counts = pd.read_csv(p_counts)
df_counts.head()
```

```python
df_ci = pd.read_csv(p_ci)
df_ci.head()
```

```python
df_merge = pd.merge(df_ci[["class", "deviation"]], df_counts, on=["class"], how="left")
df_merge["count_dev"] = df_merge["count"] * (1 - df_merge["deviation"])
df_merge
```

```python
count_ly = df_merge[df_merge["class"] == "Lycopodium"]
count_ly = count_ly.rename(columns={"count": "count_ly", "deviation" : "deviation_ly", "count_dev": "count_ly_dev"})
count_ly.drop(columns=["class"], inplace=True)
count_ly.head()
```

```python
df_merge2 = pd.merge(df_merge, count_ly, on=["site_year", "site", "year"])
df_merge3 = pd.merge(df_merge2, df_ly_tablets[["site", "year", "nb_ly_persample"]], on=["site", "year"], how="left")
df_merge3
```

```python
#  PAR min (utilise min count vitis and min count ly) et max (qui utilise max count vitis et lyco)


def compute_par_ci(count_taxon, count_dev_taxon, count_lyco, count_dev_lyco, tot_ly_sample):
    area = np.pi * 2.5*2.5
    proportion_scanned_min = float(tot_ly_sample) / max(count_lyco, count_dev_lyco)
    proportion_scanned_max = float(tot_ly_sample) / min(count_lyco, count_dev_lyco)

    par_taxon_min = min(count_taxon, count_dev_taxon) * proportion_scanned_min / area
    par_taxon_max= max(count_taxon, count_dev_taxon) * proportion_scanned_max / area
    par_taxon_mean = count_taxon * (float(tot_ly_sample) / count_lyco) / area 

    return par_taxon_mean, par_taxon_min, par_taxon_max


df_merge3[["par_mean", "par_min", "par_max"]] = df_merge3.apply(
    lambda r: compute_par_ci(r["count"],r["count_dev"],r["count_ly"],r["count_ly_dev"],r["nb_ly_persample"]),axis=1,result_type="expand")

df_merge3
```
